```python
## We will calculate the team form over the last 5 matches 
team_history={}

home_win_rate=[]

away_win_rate=[]

home_avg_goals=[]

away_avg_goals=[]

home_avg_conceded=[]

away_avg_conceded=[]
```

```python
def get_team_stats(history):
    if len(history)==0:
        return 0.5,0,0
    
    
    wins = 0
    goals_scored = 0
    goals_conceded = 0

    for match in history:
        wins += match['win']
        goals_scored += match['scored']
        goals_conceded += match['conceded']

    n = len(history)

    return (
        wins/n,
        goals_scored/n,
        goals_conceded/n
    )
```

```python
for idx,row in df.iterrows():

    home=row['home_team']
    away=row['away_team']


    home_history=team_history.get(home,[])

    away_history=team_history.get(away,[])


    home_last5=home_history[-5:]

    away_last5=away_history[-5:]


    h_win,h_goal,h_con=get_team_stats(home_last5)

    a_win,a_goal,a_con=get_team_stats(away_last5)


    home_win_rate.append(h_win)

    away_win_rate.append(a_win)

    home_avg_goals.append(h_goal)

    away_avg_goals.append(a_goal)

    home_avg_conceded.append(h_con)

    away_avg_conceded.append(a_con)


    #----------------------------------

    home_win=1 if row['home_score']>row['away_score'] else 0

    away_win=1 if row['away_score']>row['home_score'] else 0


    home_match={

        'win':home_win,

        'scored':row['home_score'],

        'conceded':row['away_score']

    }


    away_match={

        'win':away_win,

        'scored':row['away_score'],

        'conceded':row['home_score']

    }


    team_history.setdefault(home,[]).append(home_match)

    team_history.setdefault(away,[]).append(away_match)
```

```python
## NOw we will add elo ratings of the team which will boost the models prediction accuracy 

elo = {}

home_elo = []
away_elo = []
```

```python
## I will take a intermediate value of K=30 not too large not too small 

K = 30 

for idx,row in df.iterrows():
    home = row['home_team']
    away = row['away_team']

    home_rating = elo.get(home,1500)
    away_rating = elo.get(away,1500)

    home_elo.append(home_rating)
    away_elo.append(away_rating)

    expected_home = 1/(1+10**((away_rating-home_rating)/400))
    expected_away = 1 - expected_home

    if row['home_score']>row['away_score']:
        actual_home = 1
        actual_away = 0
    elif row['home_score']==row['away_score']:
        actual_home = 0.5
        actual_away = 0.5
    else:
        actual_home = 0
        actual_away = 1
    
    new_home = home_rating + K*(actual_home-expected_home)
    new_away = away_rating + K*(actual_away-expected_away)

    elo[home] = new_home
    elo[away] = new_away

    
```

```python
for idx,row in df.iterrows():
    home = row['home_team']
    away = row['away_team']

    home_hist = team_history2.get(home,[])[-5:]
    away_hist = team_history2.get(away,[])[-5:]

    if len(home_hist)==0:
        home_goal_diff_form.append(0)
    else:
        gd = sum(home_hist)/len(home_hist)
        home_goal_diff_form.append(gd)

    if len(away_hist)==0:
        away_goal_diff_form.append(0)
    else:
        gd = sum(away_hist)/len(away_hist)
        away_goal_diff_form.append(gd)

    
    current_home_gd=row['home_score']-row['away_score']
    current_away_gd=row['away_score']-row['home_score']

    team_history2.setdefault(home,[]).append(current_home_gd)
    team_history2.setdefault(away,[]).append(current_away_gd)
```

```python
def create_match_features(home_team,away_team):
    home = team_stats[home_team]
    away = team_stats[away_team]

    features_dict = {
        'home_win_rate':home['win_rate'],
        'away_win_rate' : away['win_rate'],
        'home_avg_goals' : home['avg_goals'],
        'away_avg_goals' : away['avg_goals'],
        'home_avg_conceded' : home['avg_conceded'],
        'away_avg_conceded' : away['avg_conceded'],
        'win_rate_diff' : home['win_rate'] - away['win_rate'],
        'avg_goals_diff' : home['avg_goals'] - away['avg_goals'],
        'avg_conceded_diff' : home['avg_conceded'] - away['avg_conceded'],
        'home_advantage' : 0,
        'is_friendly' : 0,
        'is_worldcup' : 1,
        'is_qualifier' : 0,
        'home_elo' : home['elo'],
        'away_elo' : away['elo'],
        'elo_diff' : home['elo'] - away['elo'],
        'elo_diff_squared' : (home['elo'] - away['elo'])**2,
        'elo_gap' : abs(home['elo'] - away['elo']),
        'form_gap' : abs(home['win_rate'] - away['win_rate'])
    }

    return pd.DataFrame([features_dict])[features]
```

```python
def predict_match(home_team, away_team):

    X = create_match_features(home_team, away_team)

    probs = xgb.predict_proba(X)[0]

    return {
        "away_win": float(probs[0]),
        "draw": float(probs[1]),
        "home_win": float(probs[2])
    }
```

```python
def simulate_match(home_team, away_team):

    X = create_match_features(
        home_team,
        away_team
    )

    probs = xgb.predict_proba(X)[0]

    outcome = np.random.choice(
        [0,1,2],
        p=probs
    )

    return outcome
```

```python
for i in range(10):
    print(
        simulate_match(
            "Brazil",
            "France"
        )
    )
```

```python
def simulate_knockout(team1, team2):

    result = simulate_match(team1, team2)

    if result == 2:
        return team1

    elif result == 0:
        return team2

    else:
        probs = predict_match(team1, team2)

        p = np.array([
            probs['home_win'],
            probs['away_win']
        ], dtype=np.float64)

        p /= p.sum()

        winner = np.random.choice(
            [team1, team2],
            p=p
        )

    return winner
```

```python
def run_tournament(teams):

    current_round = teams.copy()

    while len(current_round) > 1:

        winners = []

        for i in range(0, len(current_round), 2):

            winner = simulate_knockout(
                current_round[i],
                current_round[i+1]
            )

            winners.append(winner)

        current_round = winners

    return current_round[0]
```
